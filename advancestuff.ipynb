```python
# 70% ka split index nikalna
split_idx = int(len(finaldf) * 0.70)

# Pehla 70% Train aur aakhiri 30% Test
train_df = finaldf.iloc[:split_idx].copy()
test_df = finaldf.iloc[split_idx:].copy()

# Check karne ke liye shape aur date range print karein
print("Train shape:", train_df.shape, "| Dates:", train_df['Date'].min(), "to", train_df['Date'].max())
print("Test shape: ", test_df.shape,  "| Dates:", test_df['Date'].min(),  "to", test_df['Date'].max())
```

```python
import pandas as pd
import numpy as np

def add_7day_customer_features(df):
    df = df.copy()
    df['_orig_order'] = np.arange(len(df))
    
    # Store aur Date ke hisaab se sort karein
    df = df.sort_values(['Store', 'Date']).reset_index(drop=True)
    
    # 1. Pichle 7 calendar days ka average (current date excluded)
    rolling_avg = (
        df.groupby('Store')
          .rolling('7D', on='Date', closed='left')['Customers']
          .mean()
          .reset_index(level=0, drop=True)
    )
    df['CUST&DAYAVG'] = rolling_avg.values
    
    # 2. ind1 se ind7 indicators
    lookup = df[['Store', 'Date']].copy()
    for k in range(1, 8):
        temp = lookup.copy()
        temp['Date'] = temp['Date'] + pd.Timedelta(days=k)
        temp[f'ind{k}'] = 1
        
        df = df.merge(temp, on=['Store', 'Date'], how='left')
        df[f'ind{k}'] = df[f'ind{k}'].fillna(0).astype(int)
        
    # Original order restore karein
    df = df.sort_values('_orig_order').drop(columns=['_orig_order']).reset_index(drop=True)
    return df

# Dono par alag-alag apply karein
train_df = add_7day_customer_features(train_df)
test_df = add_7day_customer_features(test_df)
```

```python
train_df.head()
```

```python
for d in [train_df, test_df]:
    d['day_of_month'] = d['Date'].dt.day
    d['day_of_year'] = d['Date'].dt.dayofyear
    d['month_of_year'] = d['Date'].dt.month
```

```python
for d in [train_df, test_df]:
    # Total months ka difference nikalna
    d['CompetitionExistMonth'] = (
        (d['Date'].dt.year - d['CompetitionOpenSinceYear']) * 12 +
        (d['Date'].dt.month - d['CompetitionOpenSinceMonth'])
    )
    
    # Jahan NaN ho use 0 karein, aur agar competition future mein khula hai (negative value) toh use bhi 0 karein
    d['CompetitionExistMonth'] = d['CompetitionExistMonth'].fillna(0).clip(lower=0).astype(int)
```

```python
# Train set ki max distance ke hisaab se ek bahut badi value decide karein (e.g., 100,000 meters = 100 km)
# Rossmann dataset mein max CompetitionDistance lagbhag 75,860 hoti hai, isliye 100,000 ya 200,000 perfect hai
large_distance = 100000.0

for d in [train_df, test_df]:
    # 1. Jahan CompetitionExistMonth 0 hai, wahan CompetitionDistance ko bahut bada kar dein
    d.loc[d['CompetitionExistMonth'] == 0, 'CompetitionDistance'] = large_distance
    
    # 2. Agar phir bhi kisi row mein CompetitionDistance NaN bacha ho, toh use bhi large_distance se fill kar dein
    d['CompetitionDistance'] = d['CompetitionDistance'].fillna(large_distance)
```

```python
import pandas as pd

# 1. Assortment: Ordinal Mapping (basic < extra < extended)
assortment_map = {'a': 1, 'b': 2, 'c': 3}
train_df['Assortment'] = train_df['Assortment'].map(assortment_map)
test_df['Assortment'] = test_df['Assortment'].map(assortment_map)

# 2. StateHoliday cleaning: Rossmann dataset mein 0 (int) aur '0' (str) mix hote hain
train_df['StateHoliday'] = train_df['StateHoliday'].astype(str)
test_df['StateHoliday'] = test_df['StateHoliday'].astype(str)

# 3. StoreType aur StateHoliday: One-Hot Encoding (Nominal features)
train_df = pd.get_dummies(train_df, columns=['StoreType', 'StateHoliday'], drop_first=True, dtype=int)
test_df = pd.get_dummies(test_df, columns=['StoreType', 'StateHoliday'], drop_first=True, dtype=int)

# 4. Train aur Test ke columns ko exact match (align) karna
# (Agar test_df mein koi holiday category, jaise Christmas 'c', missing ho toh yeh uska column 0 se bana dega)
train_df, test_df = train_df.align(test_df, join='left', axis=1, fill_value=0)
```

```python
target_cols = ['Sales', 'Customers']

# 1. Features (X_train aur X_test) - dono targets ko drop karein
X_train = train_df.drop(columns=target_cols)
X_test = test_df.drop(columns=target_cols)

# 2. Targets alag-alag (2 separate CatBoost models train karne ke liye)
y_train_sales = train_df['Sales']
y_test_sales = test_df['Sales']

y_train_cust = train_df['Customers']
y_test_cust = test_df['Customers']
```

```python
X_train['CUST&DAYAVG'] = X_train['CUST&DAYAVG'].fillna(-1)
X_test['CUST&DAYAVG'] = X_test['CUST&DAYAVG'].fillna(-1)
```

```python
from catboost import CatBoostRegressor
from sklearn.metrics import root_mean_squared_error, mean_absolute_error
import numpy as np





model = CatBoostRegressor(
    iterations=1500,
    learning_rate=0.8,
    depth=8,
    loss_function='RMSE',
    eval_metric='RMSE',
    cat_features=['Store'],   # Store ID ko categorical treat karne ke liye
    random_seed=42,
    verbose=200               # Har 200 trees par progress dikhayega
)

# 3. Model Train karein
model.fit(
    X_train, y_train_cust,
    eval_set=(X_test, y_test_cust),
    early_stopping_rounds=300, # Agar 100 iterations tak test error kam nahi hua toh ruk jayega
    use_best_model=True
)

# 4. Prediction aur Evaluation
preds = model.predict(X_test)
# Sales kabhi negative nahi ho sakti, isliye 0 par clip karein
preds = np.clip(preds, 0, None)

rmse = root_mean_squared_error(y_test_cust, preds)
mae = mean_absolute_error(y_test_cust, preds)
print(f"Test RMSE: {rmse:.2f} | Test MAE: {mae:.2f}")
```
